```python
with open("finaldata.txt", encoding = 'utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [ c for c in chunks if len(c)>3] #it will help us to get rid of empty chunks
```

```python
def parser_chunk(chunk):
   
    chunk =  chunk.strip()
    sep_chunk = chunk.split("\n")
    username = sep_chunk[0]
        
    # Extracts the numeric value from a string like "1,234 posts" by removing text and commas, then converts it to an integer
    no_of_posts = int(sep_chunk[1].split("post")[0].replace(",","")) 
    
    ## Extracts the numeric part of followers by removing text ("followers"), commas, and K/M symbols, then converts it to float for further processing
    no_of_followers = float(sep_chunk[2].split("follower")[0].replace(",","").replace("K","").replace("M",""))
    
    #This logic ensures that values like 1.5K or 2.3M are correctly converted into full numbers without losing precision.
    if("K" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000)
    elif("M" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000000)
    else:
        no_of_followers = int(no_of_followers)

    #We perform the same logic for following as we did in followers
    no_of_followings = float(sep_chunk[3].split("following")[0].replace(",","").replace("K","").replace("M",""))
    if("K" in sep_chunk[3]):
        no_of_followings = int(no_of_followings * 1000)
    elif("M" in sep_chunk[3]):
        no_of_followings = int(no_of_followings * 1000000)
    else:
        no_of_followings = int(no_of_followings)
        
    name =  sep_chunk[4]
    if (len(sep_chunk))>6:
        type_of_page = sep_chunk[5]
        bio = "\n".join(sep_chunk[6:])
    else:
        type_of_page = "Unknow"
        bio = ""
            
    #print(username, no_of_posts,no_of_followers,no_of_followings,name,type_of_page,bio,sep = "\n")
    return {"username":username, "no_of_posts":no_of_posts,"no_of_followers":no_of_followers,"no_of_followings":no_of_followings,
            "name":name,"type_of_page":type_of_page,"bio":bio}
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parser_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json 
s = json.dumps(all_chunks, indent = 4)
with open("data.txt","w") as fd:
    fd.write(s)
```

# WHO HAS MAXIMUM POSTS?

Let's write the code to find the max posts

```python
max = 0
for chunk in all_chunks:
    if(max<chunk['no_of_posts']):
        max = chunk['no_of_posts']
        chunk_with_max_posts = chunk
print(chunk_with_max_posts['username'])
        
```

# How many categories (Digital creators, Non profit foundation etc do we have) of how many people do we have?

Let's write the code for categories of page creators

```python
category_count = {}
for chunk in all_chunks:
    page_type = chunk['type_of_page'].lower()
    ## Get current count of page_type (default 0 if not present) and add 1 (Agar mila → +1, nahi mila → 0 se start)
    category_count[page_type] = category_count.get(page_type,0)+1 
print(category_count)
```
